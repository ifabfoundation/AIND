# Assessing MCI Age Ranges

This script can be used to estimate the age range at which that patient (or an average set of patients) develops a condition of MCI (Mild Cognitive Impairment)

```python
import numpy as np
from leaspy import Leaspy, IndividualParameters, __watermark__

base_path = '../'
data_path = base_path + 'datasets/cognitive_scores.csv'
model_path = base_path + 'weights/experiment_1000_1000_9_27_19_38/'
parameters = base_path + 'weights/experiment_1000_1000_9_27_19_38/'
```

```python
def assesing_mci_age(values, timepoints):
    # Get MMSE estimated values
    MMSE_mean = values['average'][:, 0] 

    # Obtain the age range in which the MMSE is between 0.1 and 0.23 (23 and 27)
    MCI_index = np.where((MMSE_mean >= 0.1) & (MMSE_mean<= 0.23 ))
    ageMCI_a = timepoints[MCI_index[0][0]]
    ageMCI_b = timepoints[MCI_index[0][-1]]

    print('the mean population shifts to MCI (MMSE in [0.1 : 0.23]) at age: ' + str(round(ageMCI_a, 2)) + ' - ' + str(round(ageMCI_b, 2)))
```

```python
# Load model
leaspy = Leaspy.load(model_path + 'model_parameters.json')
```

```python
# Load parameters
ip = IndividualParameters.load(parameters + 'average_parameters.json')

timepoints = np.linspace(55, 130, 100)
values = leaspy.estimate({ip.to_dataframe().index.values[0]: timepoints}, ip)

assesing_mci_age(values, timepoints)
```
